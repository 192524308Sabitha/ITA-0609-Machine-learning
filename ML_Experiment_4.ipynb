```python

# EXPERIMENT 4: ARTIFICIAL NEURAL NETWORK
# BACKPROPAGATION ALGORITHM
# DATASET 2: LOAN APPROVAL PREDICTION

import pandas as pd
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Step 1: Create Dataset 2
data = {
    'Income': [
        'High', 'High', 'Medium', 'Medium',
        'Low', 'Low', 'High', 'Medium',
        'Low', 'High'
    ],
    'Credit Score': [
        'Good', 'Good', 'Good', 'Average',
        'Poor', 'Average', 'Average', 'Good',
        'Poor', 'Good'
    ],
    'Employment': [
        'Permanent', 'Permanent', 'Permanent',
        'Permanent', 'Temporary', 'Temporary',
        'Permanent', 'Permanent', 'Temporary',
        'Permanent'
    ],
    'Property': [
        'Yes', 'No', 'Yes', 'No', 'No',
        'Yes', 'Yes', 'Yes', 'No', 'Yes'
    ],
    'Loan Approved': [
        'Yes', 'Yes', 'Yes', 'Yes', 'No',
        'No', 'Yes', 'Yes', 'No', 'Yes'
    ]
}

df = pd.DataFrame(data)

print("LOAN APPROVAL DATASET:")
display(df)

# Step 2: Separate input and output
X = df[['Income', 'Credit Score', 'Employment', 'Property']]
y = df['Loan Approved']

# Step 3: Encode categorical values
encoder = OrdinalEncoder()

X_encoded = encoder.fit_transform(X)

# Encode target values
target_encoder = LabelEncoder()
y_encoded = target_encoder.fit_transform(y)

# Step 4: Scale input features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)

# Step 5: Create ANN model
model = MLPClassifier(
    hidden_layer_sizes=(5,),
    activation='relu',
    solver='adam',
    max_iter=3000,
    random_state=42
)

# Step 6: Train model using Backpropagation
model.fit(X_scaled, y_encoded)

print("\nANN MODEL TRAINED SUCCESSFULLY!")

# Step 7: Create new sample
new_sample = pd.DataFrame({
    'Income': ['Medium'],
    'Credit Score': ['Good'],
    'Employment': ['Permanent'],
    'Property': ['Yes']
})

print("\nNEW SAMPLE:")
display(new_sample)

# Step 8: Encode and scale new sample
new_encoded = encoder.transform(new_sample)
new_scaled = scaler.transform(new_encoded)

# Step 9: Predict loan approval
prediction = model.predict(new_scaled)

predicted_result = target_encoder.inverse_transform(prediction)

print("\nPREDICTED LOAN STATUS:")
print(predicted_result[0])

print("\nEXPERIMENT 4 COMPLETED SUCCESSFULLY!")
```
