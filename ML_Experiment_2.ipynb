```python

# EXPERIMENT 2: CANDIDATE ELIMINATION
# Dataset 1: Play Tennis

import pandas as pd
from itertools import product

# Step 1: Create the dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny',
            'Rainy', 'Sunny', 'Cloudy', 'Sunny'],

    'Temperature': ['Warm', 'Warm', 'Cold', 'Warm',
                    'Warm', 'Warm', 'Warm', 'Cold'],

    'Humidity': ['Normal', 'High', 'High', 'High',
                 'Normal', 'Normal', 'Normal', 'High'],

    'Wind': ['Strong', 'Strong', 'Strong', 'Strong',
             'Weak', 'Weak', 'Strong', 'Weak'],

    'Water': ['Warm', 'Warm', 'Warm', 'Cool',
              'Warm', 'Warm', 'Warm', 'Cool'],

    'Forecast': ['Same', 'Same', 'Change', 'Same',
                 'Same', 'Same', 'Same', 'Change'],

    'PlayTennis': ['Yes', 'Yes', 'No', 'Yes',
                   'No', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)

print("TRAINING DATASET:")
display(df)

# Step 2: Separate attributes and target
attributes = [
    'Sky', 'Temperature', 'Humidity',
    'Wind', 'Water', 'Forecast'
]

X = df[attributes].values.tolist()
y = df['PlayTennis'].tolist()

# Step 3: Generate possible hypotheses
domains = [
    sorted(df[attr].unique().tolist())
    for attr in attributes
]

possible_values = [domain + ['?'] for domain in domains]

all_hypotheses = list(product(*possible_values))

# Step 4: Check whether a hypothesis covers an example
def covers(hypothesis, example):
    return all(
        h == '?' or h == value
        for h, value in zip(hypothesis, example)
    )

# Step 5: Find hypotheses consistent with all examples
version_space = []

for hypothesis in all_hypotheses:
    consistent = True

    for example, target in zip(X, y):
        prediction = covers(hypothesis, example)

        if target == 'Yes' and not prediction:
            consistent = False
            break

        if target == 'No' and prediction:
            consistent = False
            break

    if consistent:
        version_space.append(hypothesis)

# Step 6: Display results
print("\nCANDIDATE ELIMINATION RESULTS")

if len(version_space) == 0:
    print("Final Version Space: EMPTY")
    print("No hypothesis is consistent with all training examples.")

    print("\nReason:")
    print("The positive examples require a general hypothesis")
    print("that also covers a negative example.")

else:
    print("Number of consistent hypotheses:", len(version_space))

    print("\nCONSISTENT HYPOTHESES:")
    for h in version_space:
        print(list(h))

    # Find maximally specific and maximally general boundaries
    def more_general_or_equal(h1, h2):
        return all(
            a == '?' or a == b
            for a, b in zip(h1, h2)
        )

    S = [
        h for h in version_space
        if not any(
            h != other and more_general_or_equal(h, other)
            for other in version_space
        )
    ]

    G = [
        h for h in version_space
        if not any(
            h != other and more_general_or_equal(other, h)
            for other in version_space
        )
    ]

    print("\nSPECIFIC BOUNDARY S:")
    for h in S:
        print(list(h))

    print("\nGENERAL BOUNDARY G:")
    for h in G:
        print(list(h))
```
